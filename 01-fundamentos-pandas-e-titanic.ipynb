# Minha primeira tabela :)

```python
import pandas as pd

dados = {
    'Cor': ['Amarela', 'Verde', 'Amarela', 'Verde'],
    'Textura': ['Macia', 'Dura', 'Macia', 'Macia'],
    'Doce': ['Sim', 'Não', 'Sim', 'Sim']
}

tabela_mangas = pd.DataFrame(dados)
display(tabela_mangas)
```

# //Consultas

```python
mangas_doces = tabela_mangas[tabela_mangas['Doce'] == 'Sim']
display(mangas_doces)
```

```python
def avaliar_manga(linha):
    saida = ''
    if linha['Doce'] == 'Sim' and linha['Textura'] == 'Macia':
        saida = 'Com certeza!'
    else:
        saida = 'Deixar no mercado'

    return saida

tabela_mangas['Comprar?'] = tabela_mangas.apply(avaliar_manga, axis=1)
display(tabela_mangas)
```
